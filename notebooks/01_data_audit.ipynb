```python
import pandas as pd

df = pd.read_csv("../data/raw/HI-Small_Trans.csv")

print(f"Total rows: {len(df):,}")
print(f"Columns: {list(df.columns)}")
print(f"\nIs Laundering value counts:")
print(df["Is Laundering"].value_counts())
print(f"\nIllicit count: {df['Is Laundering'].sum():,}")
print(f"Illicit %: {df['Is Laundering'].mean()*100:.3f}%")
```

```python
import hashlib
import json
from pathlib import Path

def sha256_file(path):
    h = hashlib.sha256()
    with open (path, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b''):
            h.update(chunk)
    return h.hexdigest()

raw_path = Path("../data/raw/HI-Small_Trans.csv")
hash_value = sha256_file(raw_path)

manifest = {
    "HI-Small_Trans.csv": {
        "sha256": hash_value,
        "rows": 5_078_345,
        "illicit": 5_177,
        "illicit_pct": 0.102,
        "licence": "CDLA-Sharing-1.0",
        "source": "https://www.kaggle.com/datasets/ealtman2019/ibm-transactions-for-anti-money-laundering-aml"
    }
}

manifest_path = Path("../data/manifest.json")
with open(manifest_path, "w") as f:
    json.dump(manifest, f, indent=2)

print(f"SHA-256: {hash_value}")
print("manifest.json written successfully")
```

```python
# Save audited data with parsed Timestamp
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# Assert no NaN in critical columns
critical_cols = ['Timestamp', 'Account', 'Account.1',
                 'Amount Received', 'Amount Paid', 'Is Laundering']
for col in critical_cols:
    assert df[col].isna().sum() == 0, f"NaN found in {col}"
    print(f" {col}: no NaN")

# Save parquet
import pathlib
pathlib.Path('../data/processed').mkdir(parents=True, exist_ok=True)
df.to_parquet('../data/processed/01_audit.parquet', index=False)
print("\n01_audit.parquet saved to data/processed/")
print(f"Shape: {df.shape}")
```
